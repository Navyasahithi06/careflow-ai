```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
patients = pd.read_csv("../datasets/cleaned/patients_cleaned.csv")
appointments = pd.read_csv("../datasets/cleaned/appointments_cleaned.csv")
er = pd.read_csv("../datasets/cleaned/hospital_er_cleaned.csv")
patient_flow = pd.read_csv("../datasets/cleaned/patient_flow_cleaned.csv")
doctors = pd.read_csv("../datasets/cleaned/doctors_cleaned.csv")
departments = pd.read_csv("../datasets/cleaned/departments_cleaned.csv")
```

```python
print(er.columns)
```

```python
er["Department Referral"].value_counts().plot(kind="bar", figsize=(10,5))

plt.title("Patients by Department")
plt.xlabel("Department")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(er["Patient Waittime"], bins=20)

plt.title("Patient Waiting Time Distribution")
plt.xlabel("Waiting Time")
plt.ylabel("Number of Patients")

plt.show()
```

```python
appointments["status"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.title("Appointment Status")

plt.ylabel("")

plt.show()
```

```python
doctors["availability"].value_counts().plot(kind="bar")

plt.title("Doctor Availability")

plt.xlabel("Status")

plt.ylabel("Doctors")

plt.show()
```

```python
patients["Condition"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Patient Conditions")

plt.xlabel("Condition")

plt.ylabel("Patients")

plt.xticks(rotation=45)

plt.show()
```
