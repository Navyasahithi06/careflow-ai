```python
import pandas as pd
import numpy as np
```

```python
patients = pd.read_csv("../datasets/hospital_patient/Hospital Patient .csv")

appointments = pd.read_csv("../datasets/appointments/appointments.csv")

er = pd.read_csv("../datasets/hospital_er/Hospital ER_Data.csv")

patient_flow = pd.read_csv("../datasets/patient_flow/healthcare_analytics_patient_flow_data.csv")

doctors = pd.read_csv("../datasets/doctors/doctors.csv")

departments = pd.read_csv("../datasets/departments/departments.csv")
```

```python
patients.columns = patients.columns.str.strip()
appointments.columns = appointments.columns.str.strip()
er.columns = er.columns.str.strip()
patient_flow.columns = patient_flow.columns.str.strip()
doctors.columns = doctors.columns.str.strip()
departments.columns = departments.columns.str.strip()
```

```python
print("Patients")
print(patients.isnull().sum())

print("\nAppointments")
print(appointments.isnull().sum())

print("\nHospital ER")
print(er.isnull().sum())

print("\nPatient Flow")
print(patient_flow.isnull().sum())

print("\nDoctors")
print(doctors.isnull().sum())

print("\nDepartments")
print(departments.isnull().sum())
```

```python
patients = patients.drop_duplicates()
appointments = appointments.drop_duplicates()
er = er.drop_duplicates()
patient_flow = patient_flow.drop_duplicates()
doctors = doctors.drop_duplicates()
departments = departments.drop_duplicates()
```

```python
print("Patients :", patients.duplicated().sum())
print("Appointments :", appointments.duplicated().sum())
print("Hospital ER :", er.duplicated().sum())
print("Patient Flow :", patient_flow.duplicated().sum())
print("Doctors :", doctors.duplicated().sum())
print("Departments :", departments.duplicated().sum())
```

```python
print(patients.dtypes)

print(appointments.dtypes)

print(er.dtypes)

print(patient_flow.dtypes)

print(doctors.dtypes)

print(departments.dtypes)
```

```python
patients.to_csv("../datasets/cleaned/patients_cleaned.csv", index=False)
appointments.to_csv("../datasets/cleaned/appointments_cleaned.csv", index=False)
er.to_csv("../datasets/cleaned/hospital_er_cleaned.csv", index=False)
patient_flow.to_csv("../datasets/cleaned/patient_flow_cleaned.csv", index=False)
doctors.to_csv("../datasets/cleaned/doctors_cleaned.csv", index=False)
departments.to_csv("../datasets/cleaned/departments_cleaned.csv", index=False)

print("All cleaned datasets saved successfully!")
```
