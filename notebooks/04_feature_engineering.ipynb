```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
er = pd.read_csv("../datasets/cleaned/hospital_er_cleaned.csv")
```

```python
print(er.columns)
```

```python
er["Patient Admission Date"] = pd.to_datetime(er["Patient Admission Date"])
```

```python
er["Admission_Year"] = er["Patient Admission Date"].dt.year
er["Admission_Month"] = er["Patient Admission Date"].dt.month
er["Admission_Day"] = er["Patient Admission Date"].dt.day
er["Admission_Weekday"] = er["Patient Admission Date"].dt.day_name()
```

```python
label_encoder = LabelEncoder()

er["Patient Gender"] = label_encoder.fit_transform(er["Patient Gender"])

er["Department Referral"] = label_encoder.fit_transform(er["Department Referral"])

er["Patient Race"] = label_encoder.fit_transform(er["Patient Race"])

er["Admission_Weekday"] = label_encoder.fit_transform(er["Admission_Weekday"])
```

```python
er = er.drop(columns=[
    "Patient Id",
    "Patient First Inital",
    "Patient Last Name",
    "Patient Admission Date",
    "Patient Satisfaction Score"
])
```
